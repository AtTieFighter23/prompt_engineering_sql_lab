# Healthcare Operations SQL Validation Notebook

This notebook supports the validation expectations for AI-assisted SQL work. Use it before accepting an AI-generated query as correct. It helps you check schema fit, row counts, date ranges, status values, joins, filters, and expected outputs.

## Validation workflow

Use these cells in order when reviewing AI-generated SQL:

1. Confirm the database connection and available tables.
2. Inspect schemas and row counts.
3. Check fixed reporting dates and valid status values.
4. Check joins before trusting multi-table results.
5. Add filters one at a time.
6. Run expected-output checks or sanity checks.

````python
import sqlite3
import pandas as pd

DB_FILE = "healthcare_operations.db"
conn = sqlite3.connect(DB_FILE)
print(f"Connected to {DB_FILE}")
````

## 1. Table list check

Start by confirming that the database contains the tables your query expects. This helps catch prompts or SQL that assume the wrong database setup.

````python
tables = pd.read_sql("""
SELECT name AS table_name
FROM sqlite_master
WHERE type = 'table'
  AND name NOT LIKE 'sqlite_%'
ORDER BY name;
""", conn)
tables
````

## 2. Schema inspection

Use the schema as the source of truth. Any table or column that does not appear here should be treated as a possible hallucination.

````python
for table_name in tables['table_name']:
    print(f"\n--- {table_name} ---")
    display(pd.read_sql(f"PRAGMA table_info({table_name});", conn))
````

## 3. Row counts

Check row counts before and after writing complex queries. If a source table has zero rows, a query that returns no rows may be behaving correctly.

````python
row_counts = []
for table_name in tables['table_name']:
    count = pd.read_sql(f"SELECT COUNT(*) AS row_count FROM {table_name};", conn).iloc[0, 0]
    row_counts.append({'table_name': table_name, 'row_count': count})
pd.DataFrame(row_counts)
````

## 6. Join coverage checks

Check whether appointment records match valid patients, providers, and clinics before trusting a multi-table query.

````python
pd.read_sql("""
SELECT
    COUNT(*) AS total_appointments,
    COUNT(p.patient_id) AS appointments_with_matching_patient,
    COUNT(pr.provider_id) AS appointments_with_matching_provider,
    COUNT(c.clinic_id) AS appointments_with_matching_clinic
FROM appointments AS a
LEFT JOIN patients AS p
    ON a.patient_id = p.patient_id
LEFT JOIN providers AS pr
    ON a.provider_id = pr.provider_id
LEFT JOIN clinics AS c
    ON a.clinic_id = c.clinic_id;
""", conn)
````

## 7. Row-count-before-and-after join checks

These joins should preserve appointment-level rows when each appointment links to one patient, provider, and clinic.

````python
pd.read_sql("""
SELECT 'appointments only' AS query_step, COUNT(*) AS row_count FROM appointments
UNION ALL
SELECT 'appointments + patients', COUNT(*)
FROM appointments AS a
JOIN patients AS p
    ON a.patient_id = p.patient_id
UNION ALL
SELECT 'appointments + patients + providers + clinics', COUNT(*)
FROM appointments AS a
JOIN patients AS p
    ON a.patient_id = p.patient_id
JOIN providers AS pr
    ON a.provider_id = pr.provider_id
JOIN clinics AS c
    ON a.clinic_id = c.clinic_id;
""", conn)
````

## 8. Filter-by-filter debugging

Use exact status values from the database. In this dataset, missed appointments are stored as `no_show`.

````python
pd.read_sql("""
SELECT 'all appointments' AS step, COUNT(*) AS row_count FROM appointments
UNION ALL
SELECT 'last 3 months', COUNT(*)
FROM appointments
WHERE appointment_date >= (SELECT recent_3_month_start FROM reporting_context)
UNION ALL
SELECT 'last 3 months and no_show', COUNT(*)
FROM appointments
WHERE appointment_date >= (SELECT recent_3_month_start FROM reporting_context)
  AND appointment_status = 'no_show';
""", conn)
````

## 9. Expected-output check: repeat no-show patients

This validates a common healthcare lab pattern: patients with repeated no-shows in the anchored recent period.

````python
repeat_no_show_patients = pd.read_sql("""
SELECT
    p.patient_id,
    p.patient_name,
    COUNT(a.appointment_id) AS no_show_count
FROM patients AS p
JOIN appointments AS a
    ON p.patient_id = a.patient_id
WHERE a.appointment_date >= (
    SELECT recent_3_month_start
    FROM reporting_context
)
  AND a.appointment_status = 'no_show'
GROUP BY p.patient_id, p.patient_name
HAVING COUNT(a.appointment_id) >= 2
ORDER BY no_show_count DESC, p.patient_name;
""", conn)

expected_columns = ['patient_id', 'patient_name', 'no_show_count']
assert list(repeat_no_show_patients.columns) == expected_columns
repeat_no_show_patients
````

# Prompt Engineering Lab: Patients with Multiple Missed Appointments

## Task 1: Planning Statement

**Business question:** Which patients had two or more missed appointments (no-shows or cancellations) in the most recent 3-month reporting period?

**Why it matters:** Repeated missed appointments disrupt follow-up care and waste scheduled provider time. Identifying these patients lets clinics target reminders and outreach to the people most likely to fall out of care.

**Key definitions:**
- **Missed appointment:** an appointment with `appointment_status` = `'no_show'` or `'canceled'` (exact lowercase values confirmed in the validation checks). `'completed'` is excluded.
- **At-risk patient:** a patient with **2 or more** missed appointments in the recent period.
- **Recent period:** `appointment_date` from `reporting_context.recent_3_month_start` (2025-03-29) through `reporting_context.reporting_date` (2025-06-29). The dataset is fictional and historical, so `date('now')` would return no rows.

**Expected output:** one row per at-risk patient with `patient_id`, `patient_name`, `insurance_type`, `no_show_count`, `canceled_count`, and `total_missed`, sorted by `total_missed` descending, then `patient_name`.

## Task 2: Schema and Query Requirements

**Tables needed:**
- `appointments`: the source of missed-appointment records
- `patients`: patient name and insurance type
- `reporting_context`: fixed date boundaries for the recent period (does not join; used in subqueries)

**Columns needed:**
- `appointments`: `appointment_id`, `patient_id`, `appointment_date`, `appointment_status`
- `patients`: `patient_id`, `patient_name`, `insurance_type`
- `reporting_context`: `recent_3_month_start`, `reporting_date`

**Join:** `patients.patient_id = appointments.patient_id` (INNER JOIN; validation cells 15 and 17 confirm every appointment matches a patient, so no rows are lost).

**Filters:**
- `appointment_status IN ('no_show', 'canceled')`
- `appointment_date >= (SELECT recent_3_month_start FROM reporting_context)`
- `appointment_date <= (SELECT reporting_date FROM reporting_context)`

**Grouping and aggregation:**
- `GROUP BY` patient (`patient_id`, `patient_name`, `insurance_type`)
- `no_show_count` = count of `'no_show'` rows; `canceled_count` = count of `'canceled'` rows (conditional `SUM(CASE ...)`)
- `total_missed` = `COUNT(appointment_id)`
- `HAVING total_missed >= 2`

**Sorting:** `total_missed DESC`, then `patient_name ASC`.

**Final output columns:** `patient_id`, `patient_name`, `insurance_type`, `no_show_count`, `canceled_count`, `total_missed`

## Task 3: Initial Structured Prompt

```text
Task: Find patients at a healthcare network who have missed multiple appointments
recently, so clinics can follow up with them.

Schema:
- patients(patient_id, patient_name, date_of_birth, city, insurance_type)
- appointments(appointment_id, patient_id, provider_id, clinic_id,
  appointment_date, appointment_status, appointment_type)
- Relationship: patients.patient_id = appointments.patient_id

Constraints:
- A missed appointment is a no-show or a cancellation.
- Only include appointments from the last 3 months.
- Only include patients with 2 or more missed appointments.

Output: Return patient_id, patient_name, insurance_type, the number of no-shows,
the number of cancellations, and the total missed appointments, sorted by total
missed (highest first).
```

## Task 4: Initial AI-Generated SQL

**AI assistant used:** Grok (xAI), in a fresh chat with no prior context. The Task 3 prompt was submitted exactly as written.

**Unedited SQL output** (line breaks restored after copy/paste collapsed them; no tokens changed):

```sql
SELECT
    p.patient_id,
    p.patient_name,
    p.insurance_type,
    SUM(CASE WHEN LOWER(a.appointment_status) IN ('no-show', 'noshow', 'no show') THEN 1 ELSE 0 END) AS num_no_shows,
    SUM(CASE WHEN LOWER(a.appointment_status) IN ('cancellation', 'cancelled', 'canceled') THEN 1 ELSE 0 END) AS num_cancellations,
    COUNT(*) AS total_missed
FROM patients p
JOIN appointments a
    ON p.patient_id = a.patient_id
WHERE a.appointment_date >= CURRENT_DATE - INTERVAL '3 months'   -- or DATE_SUB(CURDATE(), INTERVAL 3 MONTH) in MySQL
  AND (
        LOWER(a.appointment_status) IN ('no-show', 'noshow', 'no show')
     OR LOWER(a.appointment_status) IN ('cancellation', 'cancelled', 'canceled')
  )
GROUP BY p.patient_id, p.patient_name, p.insurance_type
HAVING COUNT(*) >= 2
ORDER BY total_missed DESC;
```

**AI's own notes (summarized):** The AI acknowledged it was guessing the status spellings and the SQL dialect, and told the user to adjust the string literals and interval syntax for their database.

````python
# Task 5 evidence: run the unedited initial AI SQL exactly as generated
initial_sql = """
SELECT
    p.patient_id,
    p.patient_name,
    p.insurance_type,
    SUM(CASE WHEN LOWER(a.appointment_status) IN ('no-show', 'noshow', 'no show') THEN 1 ELSE 0 END) AS num_no_shows,
    SUM(CASE WHEN LOWER(a.appointment_status) IN ('cancellation', 'cancelled', 'canceled') THEN 1 ELSE 0 END) AS num_cancellations,
    COUNT(*) AS total_missed
FROM patients p
JOIN appointments a
    ON p.patient_id = a.patient_id
WHERE a.appointment_date >= CURRENT_DATE - INTERVAL '3 months'   -- or DATE_SUB(CURDATE(), INTERVAL 3 MONTH) in MySQL
  AND (
        LOWER(a.appointment_status) IN ('no-show', 'noshow', 'no show')
     OR LOWER(a.appointment_status) IN ('cancellation', 'cancelled', 'canceled')
  )
GROUP BY p.patient_id, p.patient_name, p.insurance_type
HAVING COUNT(*) >= 2
ORDER BY total_missed DESC;
"""

# Catch the error so Restart & Run All doesn't stop here
try:
    display(pd.read_sql(initial_sql, conn))
except Exception as e:
    print(f"Initial SQL failed in SQLite:\n{type(e).__name__}: {e}")
````

````python
# Task 5 evidence: test the AI's status matching separately from the date bug
pd.read_sql("""
SELECT
    appointment_status AS actual_value,
    COUNT(*) AS rows_in_db,
    SUM(CASE WHEN LOWER(appointment_status) IN ('no-show', 'noshow', 'no show') THEN 1 ELSE 0 END) AS matched_by_ai_no_show_list,
    SUM(CASE WHEN LOWER(appointment_status) IN ('cancellation', 'cancelled', 'canceled') THEN 1 ELSE 0 END) AS matched_by_ai_cancel_list
FROM appointments
GROUP BY appointment_status
ORDER BY rows_in_db DESC;
""", conn)
````

## Task 5: SQL Review Notes

Review of the initial AI SQL against the Task 2 plan and the validation evidence above.

| Check | Result | Notes |
|---|---|---|
| Correct tables | ✅ Pass | Uses `patients` and `appointments`. No invented tables. |
| Correct join | ✅ Pass | `patients.patient_id = appointments.patient_id`, matching the relationship reference. |
| Valid columns | ✅ Pass | Every referenced column exists in the schema (checked against the PRAGMA output). |

**What worked:** schema use, the join, and the aggregation structure. The schema and relationship sections of the prompt were effective.

**What needs improvement (all prompt gaps, not AI errors):**
1. The prompt didn't give the **exact status values** → add `'no_show'` and `'canceled'` verbatim.
